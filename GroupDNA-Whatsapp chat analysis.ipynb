Project Title:
GroupDNA - WhatsApp Group Analysis

Objective:
The objective of this project is to analyze a WhatsApp group chat
using Python fundamentals and NumPy. The project extracts useful
insights from the chat such as group activity, participant
contributions, active days and hours, word usage, response times,
silent periods, and participant personality archetypes.

Technologies Used:
- Python
- NumPy
- datetime

Key Features:
1. WhatsApp Chat Parser
2. Group Overview
3. Most Active Day and Hour
4. NumPy-based Activity Analysis
5. Top Word Frequency Analysis
6. Response Time and Silent Streak Analysis
7. Personality Archetype Detection
8. Final GroupDNA Report

Dataset:
The analysis is performed on a WhatsApp group chat export
containing messages from multiple participants.

Project Constraints:
- Only Python fundamentals and NumPy are used.
- No pandas, matplotlib, seaborn, plotly, Counter,
  defaultdict, regex, or machine learning libraries are used.
- datetime is used only where required for timestamp parsing
  and time-gap calculations.


```python
# FEATURE.1- CHAT PARSER
file_path ="/content/hostel_bois.txt"

with open(file_path, "r", encoding="utf-8") as file:
    chat = file.readlines()

print("Total lines:", len(chat))
```

```python
messages=[]
for line in chat:
    if " - " in line and ": " in line:
        parts = line.split(" - ", 1)
        date, time = parts[0].split(", ")
        Sender, message = parts[1].split(": ", 1)
        messages.append({
            "Date":date,
            "Time":time,
            "Sender":Sender,
            "Message":message})
print("parsed messages:",len(messages))
```

```python
activity_matrix=np.zeros((7,24),dtype=int)
print("Matrix shape:", activity_matrix.shape)
day_index = {
    "Monday": 0,
    "Tuesday": 1,
    "Wednesday": 2,
    "Thursday": 3,
    "Friday": 4,
    "Saturday": 5,
    "Sunday": 6
}

for msg in messages:
    date_time = datetime.strptime(
        msg["Date"] + " " + msg["Time"],
        "%d/%m/%y %H:%M"
    )

    day = date_time.strftime("%A")
    hour = date_time.hour

    activity_matrix[day_index[day]][hour] += 1

print(activity_matrix)
```

Feature 5: Word Frequency

```python
word_frequency={}
for msg in messages:
  words=msg["Message"].split()
  for word in words:
    word=word.lower()
    if word not in word_frequency:
      word_frequency[word]=1
    else:
      word_frequency[word]+=1
print("Unique Words:",len(word_frequency))
```

```python
top_words = sorted(
    word_frequency.items(),
    key=lambda item: item[1],
    reverse=True
)

print("Top 10 Words")
print("-" * 30)

for word, count in top_words[:10]:
    print(word, ":", count)
```

```python
stop_words = {
    "the", "and", "i", "to", "is", "was", "in", "how", "this","you", "a", "of", "for", "it", "on", "that", "my", "me","we", "are", "be", "have",
    "has", "he", "she", "they","with", "but", "or", "so", "do", "not", "what", "your","from", "at", "as", "an", "if", "can", "just", "all","about",
    "am", "his", "which", "everyone", "today","telling", "guys"
}

meaningful_words = {}

for word, count in word_frequency.items():
    if word not in stop_words:
        meaningful_words[word] = count

top_meaningful_words = sorted(
    meaningful_words.items(),
    key=lambda item: item[1],
    reverse=True
)

print("Top 10 Meaningful Words")
print("-" * 35)

for word, count in top_meaningful_words[:10]:
    print(word, ":", count)
```

Feature 6: Response Time & Silent Streaks

```python
first_time = datetime.strptime(
    messages[0]["Date"] + " " + messages[0]["Time"],
    "%d/%m/%y %H:%M"
)

second_time = datetime.strptime(
    messages[1]["Date"] + " " + messages[1]["Time"],
    "%d/%m/%y %H:%M"
)

gap = second_time - first_time

print("First message :", first_time)
print("Second message:", second_time)
print("Time gap     :", gap)
```

```python
time_gaps = []

for i in range(1, len(messages)):
    previous_time = datetime.strptime(
        messages[i-1]["Date"] + " " + messages[i-1]["Time"],
        "%d/%m/%y %H:%M"
    )

    current_time = datetime.strptime(
        messages[i]["Date"] + " " + messages[i]["Time"],
        "%d/%m/%y %H:%M"
    )

    gap = current_time - previous_time
    time_gaps.append(gap.total_seconds())

print("Total time gaps:", len(time_gaps))
print("First 10 gaps:", time_gaps[:10])
print("Average of the timegaps:", sum(time_gaps) / len(time_gaps))
print("Longest silent streak:", max(time_gaps))
print("Fastest reesponse gap:", min(time_gaps))
```

```python
response_times = []

for i in range(1, len(messages)):

    if messages[i]["Sender"] != messages[i-1]["Sender"]:

        previous_time = datetime.strptime(
            messages[i-1]["Date"] + " " + messages[i-1]["Time"],
            "%d/%m/%y %H:%M"
        )

        current_time = datetime.strptime(
            messages[i]["Date"] + " " + messages[i]["Time"],
            "%d/%m/%y %H:%M"
        )

        gap = (current_time - previous_time).total_seconds()

        response_times.append(gap)

print("Total response gaps:", len(response_times))
print("Average response time:",
      round(np.mean(response_times), 2), "seconds")

print("Fastest response:",
      round(np.min(response_times), 2), "seconds")

print("Longest response:",
      round(np.max(response_times), 2), "seconds")

longest_gap_seconds = np.max(time_gaps)

longest_gap_hours = longest_gap_seconds / 3600

print("Longest silent streak:",
      round(longest_gap_hours, 2), "hours")
```

Feature 7: Personality Archetype Detection

```python
from datetime import datetime
import numpy as np

# Calculate Message count for each participant

participant_message_count = {}

for msg in messages:
    Sender = msg["Sender"]

    if Sender not in participant_message_count:
        participant_message_count[Sender] = 1
    else:
        participant_message_count[Sender] += 1

#Calculate total words and average words/Message

participant_word_count = {}

for msg in messages:
    Sender = msg["Sender"]
    words = msg["Message"].split()

    if Sender not in participant_word_count:
        participant_word_count[Sender] = len(words)
    else:
        participant_word_count[Sender] += len(words)


participant_avg_words = {}

for Sender in participant_message_count:
    participant_avg_words[Sender] = (
        participant_word_count[Sender]
        / participant_message_count[Sender]
    )

# Calculate response times for each participant

participant_response_times = {}

for Sender in participants:
    participant_response_times[Sender] = []


for i in range(1, len(messages)):

    previous_Sender = messages[i - 1]["Sender"]
    current_Sender = messages[i]["Sender"]

    # Only consider a change of Sender as a response
    if previous_Sender != current_Sender:

        previous_time = datetime.strptime(
            messages[i - 1]["Date"] + " " + messages[i - 1]["Time"],
            "%d/%m/%y %H:%M"
        )

        current_time = datetime.strptime(
            messages[i]["Date"] + " " + messages[i]["Time"],
            "%d/%m/%y %H:%M"
        )

        response_time = (
            current_time - previous_time
        ).total_seconds()

        participant_response_times[current_Sender].append(
            response_time
        )

#Calculate average response time

participant_avg_response = {}

for Sender in participants:

    response_list = participant_response_times[Sender]

    if len(response_list) > 0:
        participant_avg_response[Sender] = np.mean(response_list)
    else:
        participant_avg_response[Sender] = 0

# Calculate each participant's Message percentage

participant_percentage = {}

for Sender in participant_message_count:

    participant_percentage[Sender] = (
        participant_message_count[Sender]
        / total_messages
    ) * 100

#Find each participant's most-used meaningful word

participant_word_frequency = {}

for Sender in participants:
    participant_word_frequency[Sender] = {}


for msg in messages:

    Sender = msg["Sender"]

    words = msg["Message"].lower().split()

    for word in words:

        word = word.strip(".,!?;:\"'()[]{}")

        if word == "":
            continue

        if word in stop_words:
            continue

        if word not in participant_word_frequency[Sender]:
            participant_word_frequency[Sender][word] = 1
        else:
            participant_word_frequency[Sender][word] += 1


participant_top_word = {}

for Sender in participants:

    word_data = participant_word_frequency[Sender]

    if len(word_data) == 0:
        participant_top_word[Sender] = "N/A"
    else:

        sorted_words = sorted(
            word_data.items(),
            key=lambda item: item[1],
            reverse=True
        )

        participant_top_word[Sender] = sorted_words[0][0]

# Calculate group averages

average_messages_per_person = (
    total_messages / len(participants)
)

average_words_per_Message = (
    total_words / total_messages
)


response_values = []

for Sender in participants:

    if participant_avg_response[Sender] > 0:
        response_values.append(
            participant_avg_response[Sender]
        )


if len(response_values) > 0:
    group_average_response = np.mean(response_values)
else:
    group_average_response = 0

# Assign personality archetypes

personality = {}

for Sender in participants:

    message_count = participant_message_count[Sender]

    avg_words = participant_avg_words[Sender]

    avg_response = participant_avg_response[Sender]

    message_percentage = participant_percentage[Sender]


    # Very active participant
    if message_percentage >= 25:
        archetype = "The Main Character"


    # High message count + short Messages
    elif (
        message_count >= average_messages_per_person
        and avg_words < average_words_per_message
    ):
        archetype = "The Chatterbox"


    # Long Messages
    elif avg_words >= average_words_per_message * 1.5:
        archetype = "The Storyteller"


    # Fast responder
    elif (
        avg_response > 0
        and avg_response <= group_average_response * 0.5
    ):
        archetype = "The Quick Responder"


    # Low activity
    elif message_count < average_messages_per_person * 0.5:
        archetype = "The Silent Observer"


    # Default category
    else:
        archetype = "The Balanced Member"


    personality[Sender] = archetype

# Display final personality analysis

print("=" * 75)
print("              PERSONALITY ARCHETYPE ANALYSIS")
print("=" * 75)

print()

for Sender in sorted(
    participants,
    key=lambda name: participant_message_count[name],
    reverse=True
):

    print("Participant :", Sender)
    print("Messages    :", participant_message_count[Sender])
    print("Contribution:", round(participant_percentage[Sender], 2), "%")
    print("Avg Words   :", round(participant_avg_words[Sender], 2))

    if participant_avg_response[Sender] > 0:
        print(
            "Avg Response:",
            round(participant_avg_response[Sender] / 60, 2),
            "minutes"
        )
    else:
        print("Avg Response: N/A")

    print("Top Word    :", participant_top_word[Sender])
    print("Archetype   :", personality[Sender])

    print("-" * 75)

#Summary

print()
print("=" * 75)
print("                  PERSONALITY SUMMARY")
print("=" * 75)

for Sender in sorted(
    participants,
    key=lambda name: participant_message_count[name],
    reverse=True
):

    print(
        Sender,
        "->",
        personality[Sender]
    )
```
